```python
import requests

TOKEN = 'your_ncei_api_token'
dataset = 'GHCND'
locationid = 'CITY:US360019'  # Example for New York City
startdate = '2022-01-01'
enddate = '2022-01-02'

url = f'https://www.ncei.noaa.gov/access/services/data/v1?dataset={dataset}&locationid={locationid}&startdate={startdate}&enddate={enddate}&units=standard&token={TOKEN}'

response = requests.get(url)
historical_data = response.json()

print(historical_data)
```
