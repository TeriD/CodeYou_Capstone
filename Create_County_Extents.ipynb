```python
import geopandas as gpd
import pandas as pd
```

```python
# Load the GeoJSON file
geojson_path = 'data/reference_data/KY_Counties_WGS84.geojson'
counties = gpd.read_file(geojson_path)

# Ensure the GeoDataFrame has the necessary columns
if 'NAME' not in counties.columns:
    print("GeoJSON does not have a 'name' column for counties")
else:
    # Extract bounding boxes for each county
    counties['bbox'] = counties.geometry.apply(lambda x: x.bounds)

    # Create a DataFrame with county name and bounding box extents
    extents_data = {
        'county_name': counties['NAME'],
        'minx': counties['bbox'].apply(lambda x: x[0]),
        'miny': counties['bbox'].apply(lambda x: x[1]),
        'maxx': counties['bbox'].apply(lambda x: x[2]),
        'maxy': counties['bbox'].apply(lambda x: x[3])
    }

    extents_df = pd.DataFrame(extents_data)

    # Save the DataFrame to CSV
    extents_df.to_csv('data/reference_data/county_extents.csv', index=False)

    # Or save to an SQLite database
    import sqlite3

    conn = sqlite3.connect('data/crash_data.db')
    extents_df.to_sql('county_extents', conn, if_exists='replace', index=False)
    conn.close()

    print(extents_df.head())
```
